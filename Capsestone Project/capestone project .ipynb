```python
import pandas as pd

central = pd.DataFrame(index=['mean', 'median', 'mode', 'Q1:25%', 'Q2:50%', 'Q3:75%', 'Q4:100%','IQR','1.5Rule','Lesser','Greater','min','max'], columns=data.columns)

desc = data.describe()  # compute once

for col in data.columns:
    central.loc['mean', col]   = data[col].mean()
    central.loc['median', col] = data[col].median()
    central.loc['mode', col]   = data[col].mode()[0] 
    central.loc['Q1:25%', col] = desc[col]['25%']
    central.loc['Q2:50%', col]  = desc[col]['50%']
    central.loc['Q3:75%', col]  = desc[col]['75%']
    central.loc['Q4:100%', col]  = desc[col]['max']
    central.loc['IQR', col]   = central.loc['Q3:75%', col] - central.loc['Q1:25%', col] 
    central.loc['1.5Rule', col]=1.5*central.loc['IQR', col]
    central.loc['Lesser', col]=central.loc['Q1:25%', col]-central.loc['1.5Rule', col]
    central.loc['Greater', col]=central.loc['Q3:75%', col]+central.loc['1.5Rule', col]
    central.loc['min', col]=data[col].min()
    central.loc['max', col]=data[col].max()
    

central
```

```python
## Finding outlers 

lesser = []
greater = []

for col in data.columns:
    if central.loc['min', col] < central.loc['Lesser', col]:
        lesser.append(col)

    if central.loc['max', col] > central.loc['Greater', col]:
        greater.append(col)
```

```python
# ===============================
# 4. Features & Target
# ===============================
x = data[['unit_number', 'time_in_cycles', 'OS1', 'OS2', 'OS3',
          'sensor_1','sensor_2','sensor_3','sensor_4','sensor_5',
          'sensor_6','sensor_7','sensor_8','sensor_9','sensor_10',
          'sensor_11','sensor_12','sensor_13','sensor_14','sensor_15',
          'sensor_16','sensor_17','sensor_18','sensor_19','sensor_20',
          'sensor_21']]

y = data['RUL']
```

```python
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
from statsmodels.stats.diagnostic import het_breuschpagan

# Add constant
X_const = sm.add_constant(x)

# Fit OLS model
model = sm.OLS(y, X_const).fit()
print(model.summary())
```

```python
# ===============================
# 5. Train-Test Split
# ===============================
x_train, x_test, y_train, y_test = train_test_split( x, y, test_size=0.30, random_state=0)
```

```python
# ===============================
# 6. Baseline Linear Regression
# ===============================
lr = LinearRegression()
lr.fit(x_train, y_train)

y_pred = lr.predict(x_test)
print("Linear Regression R2:", r2_score(y_test, y_pred))
```

```python
# ===============================
# 7. Multiple Models Comparison
# ===============================
models = [
    ("Linear", LinearRegression()),
    ("Ridge", Ridge()),
    ("Lasso", Lasso()),
    ("SVR", SVR()),
    ("Decision Tree", DecisionTreeRegressor()),
    ("Random Forest", RandomForestRegressor()),
    ("KNN", KNeighborsRegressor()),
    ('gbr' , GradientBoostingRegressor()),
    ('xgb' , XGBRegressor()),
    ("lgbm" , LGBMRegressor())]  


for name, model in models:
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    print(f"{name}  :", r2_score(y_test, y_pred))
```

```python
# ===============================
# 8. Feature Selection (SelectKBest)
# ===============================
kbest = SelectKBest(score_func=f_regression, k=25)
X_kbest = kbest.fit_transform(x, y)


selected_features = x.columns[kbest.get_support()]
print("Selected Features:", selected_features)
```

```python
# ===============================
# 9. Split + Scaling Function
# ===============================
def split_scalar(indep_X, dep_Y):
    X_train, X_test, y_train, y_test = train_test_split(
        indep_X, dep_Y, test_size=0.25, random_state=0
    )
    sc = StandardScaler()
    X_train = sc.fit_transform(X_train)
    X_test  = sc.transform(X_test)
    return X_train, X_test, y_train, y_test
```

```python
# ===============================
# 10. R2 Prediction Function
# ===============================
def r2_prediction(model, X_test, y_test):
    y_pred = model.predict(X_test)
    return r2_score(y_test, y_pred)
```

```python
# ===============================
# 11. Model Functions
# ===============================
def Linear_Model(X_train, y_train, X_test, y_test):
    model = LinearRegression()
    model.fit(X_train, y_train)
    return r2_prediction(model, X_test, y_test)

def SVM_Linear(X_train, y_train, X_test, y_test):
    model = SVR(kernel='linear')
    model.fit(X_train, y_train)
    return r2_prediction(model, X_test, y_test)

def SVM_RBF(X_train, y_train, X_test, y_test):
    model = SVR(kernel='rbf')
    model.fit(X_train, y_train)
    return r2_prediction(model, X_test, y_test)

def Decision_Model(X_train, y_train, X_test, y_test):
    model = DecisionTreeRegressor(random_state=0)
    model.fit(X_train, y_train)
    return r2_prediction(model, X_test, y_test)

def Random_Model(X_train, y_train, X_test, y_test):
    model = RandomForestRegressor(n_estimators=100, random_state=0)
    model.fit(X_train, y_train)
    return r2_prediction(model, X_test, y_test)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from sklearn.metrics import r2_score

# ===============================
# Wrapper functions for regressors
# ===============================
def GBR_Model(X_train, y_train, X_test, y_test):
    model = GradientBoostingRegressor()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    return r2_score(y_test, y_pred)

def XGB_Model(X_train, y_train, X_test, y_test):
    model = XGBRegressor()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    return r2_score(y_test, y_pred)

def LGBM_Model(X_train, y_train, X_test, y_test):
    model = LGBMRegressor()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    return r2_score(y_test, y_pred)

# ===============================
# Apply Models on KBest Features
# ===============================
X_train_k, X_test_k, y_train_k, y_test_k = split_scalar(X_kbest, y)

results = {
    "Linear": Linear_Model(X_train_k, y_train_k, X_test_k, y_test_k),
    "SVM Linear": SVM_Linear(X_train_k, y_train_k, X_test_k, y_test_k),
    "SVM RBF": SVM_RBF(X_train_k, y_train_k, X_test_k, y_test_k),
    "Decision Tree": Decision_Model(X_train_k, y_train_k, X_test_k, y_test_k),
    "Random Forest": Random_Model(X_train_k, y_train_k, X_test_k, y_test_k),
    "GBR": GBR_Model(X_train_k, y_train_k, X_test_k, y_test_k),
    "XGB": XGB_Model(X_train_k, y_train_k, X_test_k, y_test_k),
    "LGBM": LGBM_Model(X_train_k, y_train_k, X_test_k, y_test_k)
}

# ===============================
# Display Results
# ===============================
result_df = pd.DataFrame(results, index=["R2 Score"])

```

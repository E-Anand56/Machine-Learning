```python
import pandas as pd

uni = pd.DataFrame(index=['mean', 'median', 'mode', 'Q1:25%', 'Q2:50%', 'Q3:75%', 'Q4:100%','IQR','1.5Rule','Lesser','Greater','min','max'], columns=data.columns)

numeric_cols = data.select_dtypes(include='number').columns

uni = pd.DataFrame(
    index=['mean', 'median', 'mode', 'Q1:25%', 'Q2:50%', 'Q3:75%', 'Q4:100%',
           'IQR', '1.5Rule', 'Lesser', 'Greater', 'min', 'max'],
    columns=numeric_cols
)

for col in numeric_cols:
    uni.loc['mean', col] = data[col].mean()
    uni.loc['median', col] = data[col].median()
    uni.loc['mode', col] = data[col].mode().iloc[0]
    uni.loc['Q1:25%', col] = data[col].quantile(0.25)
    uni.loc['Q2:50%', col] = data[col].quantile(0.50)
    uni.loc['Q3:75%', col] = data[col].quantile(0.75)
    uni.loc['Q4:100%', col] = data[col].quantile(1.00)
    uni.loc['IQR', col] = uni.loc['Q3:75%', col] - uni.loc['Q1:25%', col]
    uni.loc['1.5Rule',col]=1.5*uni.loc['IQR', col]
    uni.loc['Lesser',col]=uni.loc['Q1:25%', col]-uni.loc['1.5Rule',col]
    uni.loc['Greater',col]=uni.loc['Q3:75%', col]+uni.loc['1.5Rule',col]
    uni.loc['min', col] = data[col].min()
    uni.loc['max', col] = data[col].max()

uni
```

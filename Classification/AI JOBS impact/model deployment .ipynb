```python
import pandas as pd
import pickle

# Load model
model = pickle.load(open('finalmodel_RandomForestClassifier.sav', 'rb'))

# Get user inputs
data = { 
    'Average_Salary': [float(input("Enter average salary: "))],
    'Automation_Probability_2030': [float(input("Enter automation chance (0-1): "))],
    'AI_Exposure_Index': [float(input("Enter AI exposure index (0-1): "))],
    'Tech_Growth_Factor': [float(input("Enter tech growth factor: "))],
    'Years_Experience': [int(input("Enter years of experience: "))],
    'Skill_1': [int(input("Enter skill code (e.g., 1,2,3): "))]
}

# Create DataFrame
input_data = pd.DataFrame(data)

# Map prediction to readable label in one line
readable_result = {0: 'High', 1: 'Low', 2: 'Medium'}[model.predict(input_data)[0]]

print("Predicted Risk Category:", readable_result)
```
